```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
Data={"Student":["Hema","Guru","guha","vasuki","vishnu","mithra","Seetha","kumar","kirthik","Priya","KiKi","Harini","sweet","pricila"],
      "Department":["CS","IT","python","IT","python","IT","CS","IT","CS","python","java","java","C++","C++"],
      "Marks":[85,90,78,65,95,72,88,60,95,75,67,80,65,87],
      "Attendance":[90,95,80,70,98,75,92,65,96,82,90,89,70,92]}
df=pd.DataFrame(Data)
print(df)
```

```python
average_Marks=np.mean(df["Marks"])
print("Average_Marks:",average_Marks)
```

```python
highest_Marks=np.max(df["Marks"])
print("Highest_Marks:",highest_Marks)
```

```python
lowest_Marks=np.min(df["Marks"])
print("Lowest_Marks:",lowest_Marks)
```

```python
cs_students=df[df["Department"]=="CS"]
print(cs_students)
```

```python
print("Total CS Students:",len(cs_students))
```

```python
plt.figure(figsize=(10,5))
plt.bar(df["Student"], df["Marks"])
plt.title("Student Performance Analysis")
plt.xlabel("Student name")
plt.ylabel("Marks")
plt.show()
```

```python
department_count=df["Department"].value_counts()
plt.figure(figsize=(6,6))
plt.pie(
    department_count,
    labels=department_count.index,
    autopct="%1.1f%%",
    startangle=90
 )            
plt.title("Department Wise Student Percentage")
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.plot(df["Student"], df["Marks"],marker="o")
plt.title("Student Marks Performance")
plt.xlabel("Student Name")
plt.ylabel("Marks")
plt.grid(True)
plt.show()
```

```python
department_average=df.groupby("Department")["Marks"].mean()
print(department_average)
```

```python
print("STUDENT PERFORMANCE REPORT")
print("-------------------------")
print("Total Students:",len(df))
print("Average Marks:",np.mean(df["Marks"]))
print("Highest Marks:",np.max(df["Marks"]))
print("Lowest Marks:",np.min(df["Marks"]))
print("\nDepartment Average:")
print(df.groupby("Department")
["Marks"].mean())

```

```python
df.to_csv("Student_Performance_analysis.csv",index=False)
print("CSV File Saved Successfully!")
```
